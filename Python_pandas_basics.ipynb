### Фильтры с помощью функций
Пишем функцию, которая для каждой строчки возвращает True, либо False.
В зависимости от значений.

```python
df.head()
```

```python
# .isin(['Новосибирск', 'Москва'])
#apply() — метод библиотеки pandas для Python, 
#который позволяет применить функцию к каждому элементу (в случае Series) или к каждой строке/колонке (в случае DataFrame).
df[
    df['city'].apply(lambda city: city in ['Москва', 'Новосибирск'])
]
```

```python
df.head()
```

```python
def complex_filter(row):
    """Фильтрует строки, соответствующие продажам в новогоднюю акцию"""
    date = row['date']
    city = row['city']
    sales = row['sales']
    
    if date == '2021-01-01':
        if city in ['Москва', 'Владивосток']:
            return True
        
    if date == '2021-01-02':
        if sales > 100 and city == 'Тула':
            return True
        
    return False
```

```python
df[df.apply(complex_filter, axis=1)]
```

### Фильтр с отрицанием условия
Ставим значок тильда

```python
df[~df.apply(complex_filter, axis=1)]
```

# Фильтры

```python
data = pd.read_csv('https://raw.githubusercontent.com/ahartz1/car-linear-regression/master/car_data.csv')
data.head()
```

```python
# выбрать несколько столбцов

car_stats = data.filter(items = ['Price', 'Mileage'])
car_stats.head()
```

```python
# или так

data[['Price', 'Mileage']].head()
```

### Отфильтруем строки с пробегом авто выше среднего

```python
average_level = data['Mileage'].mean()
average_level
```

```python
# строки с пробегом  больше среднего

average_level = data['Mileage'].mean()
car_stats.query(f'Mileage > {average_level}').head()
```

```python
# самый популярный способ

data[ data.Mileage > average_level ].head()
```
